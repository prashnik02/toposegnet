```python
import numpy as np
import richdem as rd
from tqdm import tqdm

splits = {
    "Train": 8490,
    "Val":   535,
    "Test":  975,
}

base = r"C:\Users\prash\toposegnet\data\high_509"

for split, n in splits.items():
    folder = f"{base}\\{split}"
    print(f"\n=== Processing {split} ({n} patches) ===")

    # input elevation memmap
    elev = np.memmap(f"{folder}\\EXTRA_elevation.dat",
                     dtype='int16', mode='r', shape=(n, 512, 512))

    # output memmaps — float32, same shape
    slope_mm     = np.memmap(f"{folder}\\DERIVED_slope.dat",
                             dtype='float32', mode='w+', shape=(n, 512, 512))
    sin_asp_mm   = np.memmap(f"{folder}\\DERIVED_sin_aspect.dat",
                             dtype='float32', mode='w+', shape=(n, 512, 512))
    cos_asp_mm   = np.memmap(f"{folder}\\DERIVED_cos_aspect.dat",
                             dtype='float32', mode='w+', shape=(n, 512, 512))

    for i in tqdm(range(n)):
        patch = elev[i].astype(np.float32)
        rd_patch = rd.rdarray(patch, no_data=-9999)

        slope   = np.array(rd.TerrainAttribute(rd_patch, attrib='slope_degrees'))
        aspect  = np.array(rd.TerrainAttribute(rd_patch, attrib='aspect'))

        aspect_rad = np.deg2rad(aspect)

        slope_mm[i]   = slope
        sin_asp_mm[i] = np.sin(aspect_rad)
        cos_asp_mm[i] = np.cos(aspect_rad)

    # flush to disk
    slope_mm.flush()
    sin_asp_mm.flush()
    cos_asp_mm.flush()
    print(f"  Saved DERIVED_slope.dat, DERIVED_sin_aspect.dat, DERIVED_cos_aspect.dat")

print("\nDone.")
```

```python
import numpy as np

slope_train = np.memmap(
    r"C:\Users\prash\toposegnet\data\high_509\Train\DERIVED_slope.dat",
    dtype='float32', mode='r', shape=(8490, 512, 512)
)

# chunk-based to avoid RAM spike
chunk_size  = 100
total_pixels = 0
mean = 0.0
M2   = 0.0

for i in range(0, 8490, chunk_size):
    chunk = slope_train[i:i+chunk_size].ravel().astype(np.float64)
    n      = len(chunk)
    total_pixels += n
    delta  = chunk - mean
    mean  += delta.sum() / total_pixels
    delta2 = chunk - mean
    M2    += (delta * delta2).sum()
    if (i // chunk_size) % 10 == 0:
        print(f"  {i}/8490...")

std = np.sqrt(M2 / total_pixels)
print(f"\nSlope stats:")
print(f"  Mean: {mean:.4f}")
print(f"  Std : {std:.4f}")
```

```python
import numpy as np
import torch
from torch.utils.data import Dataset
import albumentations as A
import pandas as pd

# ── constants ────────────────────────────────────────────────────────────────
DATA_ROOT  = r"C:\Users\prash\toposegnet\data\high_509"

SPLIT_SIZES = {"Train": 8490, "Val": 535, "Test": 975}

BAND_FILES = [
    "L1C_B1.dat",  "L1C_B2.dat",  "L1C_B3.dat",  "L1C_B4.dat",
    "L1C_B5.dat",  "L1C_B6.dat",  "L1C_B7.dat",  "L1C_B8.dat",
    "L1C_B8A.dat", "L1C_B9.dat",  "L1C_B10.dat", "L1C_B11.dat",
    "L1C_B12.dat",
]

# normalization constants (computed from train split only)
S2_SCALE   = 10000.0
ELEV_MEAN  = 683.6302;  ELEV_STD   = 1025.6890
SLOPE_MEAN =  17.9182;  SLOPE_STD  =   28.9855

# ── ablation modes ───────────────────────────────────────────────────────────
# "none"    → 13 channels
# "elev"    → 14 channels
# "optionA" → 15 channels (elev + slope, no aspect)
# "optionB" → 17 channels (elev + slope + sin_asp + cos_asp)
# "optionC" → 19 channels (optionB + lat + lon)
TERRAIN_MODES = ["none", "elev", "optionA", "optionB", "optionC"]

# ── augmentations ─────────────────────────────────────────────────────────────
def get_transforms(split, augment=True):
    if split == "Train" and augment:
        return A.Compose([
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
        ])
    return None   # no augmentation for val/test or when augment=False

# ── dataset ───────────────────────────────────────────────────────────────────
class CloudSEN12Dataset(Dataset):
    """
    Args:
        split        : "Train" | "Val" | "Test"
        terrain_mode : "none" | "elev" | "optionA" | "optionB" | "optionC"
        augment      : True/False — only applies to Train split

    Returns:
        image : float32 tensor (C, 512, 512)
                C = 13/14/15/17/19 depending on terrain_mode
        label : long tensor (512, 512), values 0-3
    """

    def __init__(self, split="Train", terrain_mode="none", augment=True):
        assert split        in SPLIT_SIZES,   f"split must be one of {list(SPLIT_SIZES)}"
        assert terrain_mode in TERRAIN_MODES, f"terrain_mode must be one of {TERRAIN_MODES}"

        self.split        = split
        self.terrain_mode = terrain_mode
        self.n            = SPLIT_SIZES[split]
        self.transforms   = get_transforms(split, augment)

        folder = f"{DATA_ROOT}\\{split}"
        shape  = (self.n, 512, 512)

        # ── spectral bands ───────────────────────────────────────────────────
        self.bands = [
            np.memmap(f"{folder}\\{f}", dtype='uint16', mode='r', shape=shape)
            for f in BAND_FILES
        ]

        # ── terrain memmaps ──────────────────────────────────────────────────
        self.elev    = None
        self.slope   = None
        self.sin_asp = None
        self.cos_asp = None
        self.lats    = None
        self.lons    = None

        if terrain_mode in ["elev", "optionA", "optionB", "optionC"]:
            self.elev = np.memmap(f"{folder}\\EXTRA_elevation.dat",
                                  dtype='int16', mode='r', shape=shape)

        if terrain_mode in ["optionA", "optionB", "optionC"]:
            self.slope = np.memmap(f"{folder}\\DERIVED_slope.dat",
                                   dtype='float32', mode='r', shape=shape)

        if terrain_mode in ["optionB", "optionC"]:
            self.sin_asp = np.memmap(f"{folder}\\DERIVED_sin_aspect.dat",
                                     dtype='float32', mode='r', shape=shape)
            self.cos_asp = np.memmap(f"{folder}\\DERIVED_cos_aspect.dat",
                                     dtype='float32', mode='r', shape=shape)

        if terrain_mode == "optionC":
            meta = pd.read_csv(f"{folder}\\metadata.csv")
            centroids = meta["proj_centroid"].str.extract(
                r'POINT \(([0-9.\-]+) ([0-9.\-]+)\)'
            )
            self.lons = centroids[0].astype(float).values  # (n,)
            self.lats = centroids[1].astype(float).values  # (n,)

        # ── label ────────────────────────────────────────────────────────────
        self.label = np.memmap(f"{folder}\\LABEL_manual_hq.dat",
                               dtype='int8', mode='r', shape=shape)

    def __len__(self):
        return self.n

    def __getitem__(self, idx):
        # ── load spectral patch → (512, 512, 13) ────────────────────────────
        image = np.stack([b[idx] for b in self.bands], axis=-1).astype(np.float32)
        label = self.label[idx].astype(np.int64)

        # ── collect terrain as extra masks for albumentations ────────────────
        # albumentations applies identical spatial transforms to all masks
        extra_masks = []

        if self.elev is not None:
            extra_masks.append(self.elev[idx].astype(np.float32))
        if self.slope is not None:
            extra_masks.append(self.slope[idx].copy())
        if self.sin_asp is not None:
            extra_masks.append(self.sin_asp[idx].copy())
            extra_masks.append(self.cos_asp[idx].copy())

        # ── augment (train only) ─────────────────────────────────────────────
        if self.transforms is not None:
            out         = self.transforms(image=image, mask=label, masks=extra_masks)
            image       = out["image"]
            label       = out["mask"]
            extra_masks = out["masks"]

        # ── normalize spectral ───────────────────────────────────────────────
        image = image / S2_SCALE   # → ~0.0 to ~2.8

        # ── normalize terrain ────────────────────────────────────────────────
        terrain_channels = []
        ptr = 0

        if self.elev is not None:
            terrain_channels.append((extra_masks[ptr] - ELEV_MEAN) / ELEV_STD)
            ptr += 1

        if self.slope is not None:
            terrain_channels.append((extra_masks[ptr] - SLOPE_MEAN) / SLOPE_STD)
            ptr += 1

        if self.sin_asp is not None:
            terrain_channels.append(extra_masks[ptr])      # sin aspect, already [-1,1]
            terrain_channels.append(extra_masks[ptr + 1])  # cos aspect, already [-1,1]
            ptr += 2

        if self.lats is not None:
            h, w  = label.shape
            lat_n = self.lats[idx] / 90.0    # normalize to [-1, 1]
            lon_n = self.lons[idx] / 180.0   # normalize to [-1, 1]
            terrain_channels.append(np.full((h, w), lat_n, dtype=np.float32))
            terrain_channels.append(np.full((h, w), lon_n, dtype=np.float32))

        # ── to tensor → (C, 512, 512) ────────────────────────────────────────
        image_t = torch.from_numpy(image.transpose(2, 0, 1))  # (13, 512, 512)

        if terrain_channels:
            terrain_t = torch.from_numpy(
                np.stack(terrain_channels, axis=0)            # (T, 512, 512)
            )
            image_t = torch.cat([image_t, terrain_t], dim=0) # (13+T, 512, 512)

        label_t = torch.from_numpy(label)                     # (512, 512)

        return image_t, label_t
```

```python
for mode, expected_c in [
    ("none", 13), ("elev", 14), ("optionA", 15),
    ("optionB", 17), ("optionC", 19)
]:
    ds = CloudSEN12Dataset("Train", terrain_mode=mode)
    img, lbl = ds[0]
    status = "OK" if img.shape[0] == expected_c else f"WRONG (got {img.shape[0]})"
    print(f"{mode:10s} → image={img.shape}, label={lbl.shape} [{status}]")
```

```python
import numpy as np
from tqdm import tqdm

label_train = np.memmap(
    r"C:\Users\prash\toposegnet\data\high_509\Train\LABEL_manual_hq.dat",
    dtype='int8', mode='r', shape=(8490, 512, 512)
)

# count pixels per class across all 8490 patches
counts = np.zeros(4, dtype=np.int64)

for i in tqdm(range(8490)):
    patch = label_train[i]
    for c in range(4):
        counts[c] += (patch == c).sum()

total = counts.sum()
print(f"\nPixel counts:")
print(f"  0 clear  : {counts[0]:>12,} ({100*counts[0]/total:.2f}%)")
print(f"  1 thick  : {counts[1]:>12,} ({100*counts[1]/total:.2f}%)")
print(f"  2 thin   : {counts[2]:>12,} ({100*counts[2]/total:.2f}%)")
print(f"  3 shadow : {counts[3]:>12,} ({100*counts[3]/total:.2f}%)")

# inverse frequency weighting: weight_c = total / (n_classes * count_c)
weights = total / (4 * counts)
weights = weights / weights.min()   # normalize so smallest weight = 1.0

print(f"\nClass weights (normalized):")
print(f"  clear  : {weights[0]:.4f}")
print(f"  thick  : {weights[1]:.4f}")
print(f"  thin   : {weights[2]:.4f}")
print(f"  shadow : {weights[3]:.4f}")
```
